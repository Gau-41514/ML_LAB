```python
import numpy as np
import pandas as pd

states = ["Walking" , "Running" , "Resting"]
symbols = ["Low","Medium","High"]
obss = [2,1,0,2]
pi = np.array([0.5,0.3,0.2])
A = np.array([
    [0.6,0.3,0.1],
    [0.2,0.5,0.3],
    [0.3,0.3,0.5]
])
B = np.array([
    [0.6,0.3,0.1],
    [0.1,0.3,0.6],
    [0.7,0.2,0.1]
])
print("-------------A----------------")
print(A)
print("-------------B----------------")
print(B)
print("-------------pi----------------")
print(pi)
```

```python
p = len(obss)
q = len(states)
X = np.zeros((p,q))
print(X)
```

```python
#obss = high
X[0][0] = pi[0]*B[0][2]
X[0][1] = pi[1]*B[1][2]
X[0][2] = pi[2]*B[2][2]
print(X)
```

```python
#obss = mid
X[1][0] = (X[0][0]*A[0][0] + X[0][1]*A[1][0] + X[0][2]*A[2][0]) * B[0][1]
X[1][1] = (X[0][0]*A[0][1] + X[0][1]*A[1][1] + X[0][2]*A[2][1]) * B[1][1]
X[1][2] = (X[0][0]*A[0][2] + X[0][1]*A[1][2] + X[0][2]*A[2][2]) * B[2][1]
print(X)
```

```python
#obss = low
X[2][0] = (X[1][0]*A[0][0] + X[1][1]*A[1][0] + X[1][2]*A[2][0]) * B[0][0]
X[2][1] = (X[1][0]*A[0][1] + X[1][1]*A[1][1] + X[1][2]*A[2][1]) * B[1][0]
X[2][2] = (X[1][0]*A[0][2] + X[1][1]*A[1][2] + X[1][2]*A[2][2]) * B[2][0]
print(X)
```

```python
#obss = high
X[3][0] = (X[2][0]*A[0][0] + X[2][1]*A[1][0] + X[2][2]*A[2][0]) * B[0][2]
X[3][1] = (X[2][0]*A[0][1] + X[2][1]*A[1][1] + X[2][2]*A[2][1]) * B[1][2]
X[3][2] = (X[2][0]*A[0][2] + X[2][1]*A[1][2] + X[2][2]*A[2][2]) * B[2][2]
print(X)
```

```python
Total_prob = np.sum(X[3])
print(Total_prob)
```
